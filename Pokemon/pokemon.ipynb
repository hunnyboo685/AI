# My Question
## How does the weight differ between water type and fire type Pokemons?

## Type of Study
### This is a census study because the dataset includes every single member of the Pokemon population. I can observe different types of Pokemons and their weights through the population of Pokemons in this dataset.

## Analysis: 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
pokemon = pd.read_csv("pokemon.csv")
```

```python
pokemon.head()
```

```python
pokemon.info()
```

```python
subdf = pokemon[['weight_kg']].copy()
subdf.head()
```

```python
graph = sns.histplot(subdf, x = "weight_kg")
```

```python
#use interquartile range
```

```python
s = pokemon[(pokemon["primary_type"] == "water")]
s.head()
```

```python
water_df = pokemon[pokemon["primary_type"] == "water"][
    ["primary_type", "weight_kg"]
]
water_df.head()
```

```python
fire_df = pokemon[pokemon["primary_type"] == "fire"][
    ["primary_type", "weight_kg"]
]
fire_df.head()
```

```python
print(water_df["weight_kg"].describe())
print(fire_df["weight_kg"].describe())
```

```python
#water_df Q1: 9.50   Q3: 72.50
```

```python
#fire_df Q1: 1.00         Q3: 81.23
```

```python
quantilew = 72.50 - 9.50
print(quantilew)
```

```python
quantilef = 81.23 - 1.00
print(quantilef)
```

```python
sns.boxplot(data = water_df  , x = 'weight_kg')
sns.boxplot(data = fire_df   , x = 'weight_kg')
```

```python
combo = pd.concat([water_df, fire_df])
```

```python
sns.boxplot(data = combo   , x = 'weight_kg' , y = 'primary_type')
```

# Conclusion
## The fire-types of Pokemon have a spread of weight from 1 - 81.23 with an IQR of 80.23 which is represented through the boxplots. 
## The water-type Pokemon which have have a spread of 9.50 to 72.50 and an IQR of 63 which is rperesented on the boxplots. 
## The weights of the fire and water Pokemons do not really differ rather they are similar. On the first boxplot the two types overlap while the second plot demonstrates a side-by side version where they have similar ranges and IQR showing that they are similar and don't have a significant difference.
