# What's the average danceability score between all the songs that were popular in 2023?

## This is an observational study because it is a list of the most popular songs in 2023 picked from Spotify showing that these songs are a sample and not the entire population of songs that could've been popular that year.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
music = pd.read_csv("spotify-2023.csv" , encoding = "latin-1")
```

```python
music.head()
```

```python
music.info()
```

```python
#variable : danceablity_%
```

```python
sub_df = music[['danceability_%']].copy()
sub_df.head()
```

```python
number = len(sub_df)
print("Amount:" , number)
```

```python
addition = pd.Series([sub_df], dtype="float64").sum() 
print("Sum of all songs' danceability:" , addition)
```

```python
avg = addition/number
print("Avg danceability of popular 2023 songs" , avg)
```

## Analysis: The average danceability scores of Spotify's most popular songs of 2023 is 66.97

```python
graph = sns.histplot(sub_df, x = "danceability_%")
graph.axvline(
    avg, color="red", linestyle="--", linewidth=2, label=f"Mean: {avg:.2f}"
)


plt.legend()
plt.show()
```

# Conclusion: Out of all the popular songs of 2023 given by Spotify, 66.97 was the average danceability.
