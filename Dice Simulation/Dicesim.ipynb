```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

# My Question:
## What is the probability of rolling a number greater than 3 on a die with 6 sides 1000 times?

# My Study:
### 1. The random process being simulated is rolling a die.
### 2. In one trial, either a number greater than three or three and less will be rolled on. Possible outcomes would be greater than 3 : 4, 5, 6 or 3 and under: 1, 2, 3.
### 3. I am interested in the probability of the die landing on numbers greating than 3 as an outcome. These numbers are 4,5,6. 

# My Analysis:

```python
# theoretical probability: favorable outcomes/total possible outcomes      
# empirical probability: number of times event occurred/ total number of trials
```

```python
rng = np.random.default_rng()
```

```python
rolls = rng.integers(low=1, high=7, size=1000)
print("First 20 rolls:", rolls[:20])
```

```python
targetnum = rolls > 3
```

```python
biggerthan3 = np.sum(targetnum)
simprob = biggerthan3 / len(rolls)
```

```python
print(f"Total rolls > 3: {biggerthan3}")
print(f"Empirical Probability: {simprob:.4f}")
```

### The probability of rolling a number greater than 3 from the simulation is 0.5010.

```python
theoprob = 3/6
print(theoprob)
```

### The theoretical probability of rolling a number greater than 3 is 3/6 or 0.5.

```python
trials = np.arange(1, len(rolls) + 1)
cumulative_simprob = np.cumsum(targetnum) / trials
```

```python
plt.figure(figsize=(9, 5))
plt.plot(trials, cumulative_simprob, label=f"Simulated Prob (Final: {simprob:.4f})", color="magenta", lw=1.5)
plt.axhline(y=theoprob, color="turquoise", linestyle="--", lw=2, label=f"Theoretical Prob ({theoprob})")
plt.title("Empirical vs. Theoretical Probability Over 1,000 Rolls")
plt.xlabel("Number of Rolls")
plt.ylabel("Probability")
plt.ylim(0.3, 0.7)
plt.grid(True, linestyle=":", alpha=0.6)

plt.legend()
plt.show()
```

# My Conclusion:
## The empirical probability of rolling a number above 3 in the simulation was 0.5010. However, the theoretical probability of this outcome is 0.50. Overall, the probability is 0.5. The graph above supports my answer because even though at first the cumulative/running emmpricial probability is not similar to the theoretical probability, as the number of rolls increases both probabilities align next to/at 0.5. This shows that the probability of rolling a die and getting a number greater than 3 for 1000 times is 0.5.
