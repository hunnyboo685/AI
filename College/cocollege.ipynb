# Question: How do the average SAT scores of UC Berkeley and CU Boulder compare to all post-secondary insitutions in the United States? These colleges are both public universities.

## Type of Study: This is a census study as we are calculating z-scores for specific schools from a dataset containing the entire college population in the USA as well as getting an average from the entire population. This is a dataet from the College Scorecard website set up by the U.S. Department of Education.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
college = pd.read_csv("cleaned_college.csv")
```

```python
college.head()
```

```python
college.info()
```

```python
subdf = college[['sat_average']].copy()
subdf.head()
```

```python
mean = subdf[["sat_average"]].mean()
print(mean)
```

```python
median = subdf[["sat_average"]].median()
print(median)
```

```python
graph = sns.histplot(subdf, x = "sat_average")
```

```python
stdb = subdf[["sat_average"]].std()
print(stdb)
```

```python
# average SAT score for UC Berkeley is not listed so I searched it up
```

```python
avgUCB = 1415
print(avgUCB)
```

```python
zscoreb = (avgUCB - mean)/ stdb
print(zscoreb)
```

## Z-score of UC Berkeley is 1.55

```python
boulder = college[college["school_name"] == "University of Colorado Boulder"]
sat = boulder["sat_average"].item()
print(sat) 
```

```python
zscorec = (sat - mean)/stdb
print(zscorec)
```

## Z-score of CU Boulder is 1.16

```python
avg = 1184.464802
```

```python
graph = sns.histplot(subdf, x = "sat_average")
graph.axvline(
    avg, color="red", linestyle="--", linewidth=2, label=f"Mean: {avg:.2f}"
)


plt.legend()
plt.show()
```

# Conclusion: UC Berkeley is almost 2 standard deviations(zscore: 1.55) from the mean however it is not an outlier in the distribution of college SATS in the USA. Although the average SAT scores for Berkeley are higher in the distribution it is still close to many other colleges' average SAT scores. CU boulder is a little above 1 standard deviation(zscore: 1.16) from the mean showing that it is central around the mean in the distribution and lies in the average SAT score acceptance range of many other colleges.
