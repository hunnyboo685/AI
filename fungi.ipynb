# How many poisonous mushrooms grow solitarily in the woods habitat in the dataset?

## This is an observational study because we are observing the frequency of times this specific mushroom is identified in the dataset.

```python
import pandas as pd
```

```python
import seaborn as sns
```

```python
fungi = pd.read_csv("mushrooms.csv")
```

```python
fungi.head()
```

```python
fungi.info()
```

```python
#variables are class(p = poisonous), population(y = solitarily), habitat(d = woods)  
```

```python
mushrooms = fungi[(fungi["class"] == "p") & (fungi["population"] == "y") & (fungi["habitat"] == "d")]
```

```python
count = len(mushrooms)
print("Amount:", count)
```

### 216 mushrooms are solitary-growing poisonous mushroms from the woods habitat.

```python
sub_df = mushrooms[['population', 'class', 'habitat']].copy()
```

```python
p = fungi[fungi["class"] == 'p']
```

```python
s = p[p["population"] == "y"]
s.head()
```

```python
w = s[s["habitat"] == 'd']
w.head()
```

```python
sns.countplot(w, x='class')
```
