# How many people like their steak medium rare and are over the age of 60?

### This is an observational study because I can reasonably observe and make claims about the sample group in the survey.

```python
import pandas as pd
import seaborn as sns
```

```python
food = pd.read_csv("steak.csv")
```

```python
pref = food[(food["How do you like your steak prepared?"] == "Medium rare") & (food["Age"] == ">60")]
```

```python
count = len(food)
print("This many people over 60 prefer their steak to be medium rare:", count)
```
